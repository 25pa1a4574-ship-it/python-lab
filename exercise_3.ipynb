```python
n=int(input())
for i in range(2,n):
   if(n%i== 0):
    print("it is not prime number")
    break
else:
    print("it is prime number")

```

#print the sum of the digits of given number


```python
n=int(input())
sum=0
while(n!=0):
  rem=n%10
  sum=sum+rem
  n=n//10
  print(sum)
```
