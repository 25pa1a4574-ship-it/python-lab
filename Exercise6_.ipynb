#PALINDROME OR NOT

```python
def palindrome(num):
  original=num
  rev=0
  while(num!=0):
    rem=num%10
    rev=rev*10+rem
    num=num//10
  if(rev==original):
    return True
  else:
    return False
print(palindrome(121))
print(palindrome(4574))
```

# REVERSE A STRING USSING FUNCTIONS


```python
def reverse_str(s):
  rev=""
  for i in range(len(s)-1,-1,-1):
     rev=rev+s[i]
  return rev
print(reverse_str("hello"))

```

#FUNCTION THAT RETURNS MULITIPLE

```python
def multiple_value(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_value(1,2,3,4,5)
print(f"sum is {res_sum},count is {res_count},average{res_avg}")
```
