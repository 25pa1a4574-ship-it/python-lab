#PROGRAM THAT COUNT THE NO OF VOWELS AND CONSTANTS IN A STRING

```python
a = str(input())
b = a.lower()
vowels ="aeiou"
count_c = 0;
count_v = 0;
for ch in b:
  if ch.isalpha():
    if ch in vowels:
      count_v+=1
    else:
      count_c+=1
print("no of vowels",count_v)
print("no pf constants",count_c)

```
